```python
import pandas as pd
import json 
import os
```

```python
def read_first_json_file(directory_path):
    for filename in os.listdir(directory_path):
        if filename.endswith('.json'):
            filepath = os.path.join(directory_path, filename)
            with open(filepath, 'r', encoding='utf-8') as f:
                data = json.load(f)
            return data, filename
    return None, None

data, filename = read_first_json_file('../json_file')
if data:
    df = pd.DataFrame(data)
    df.info()
```

```python
display(df.head())
```

```python
stat_df = pd.json_normalize(df['statistics'])
display(stat_df)
```

```python
df1 = pd.concat([df.drop(columns=['statistics']), stat_df], axis=1)
df1 = df1.drop(columns=['comments', 'dislikeCount',  'description']).reset_index(drop=True)
df1['likeCount'] = pd.to_numeric(df1['likeCount'], errors='coerce')
df1['datetime'] = pd.to_datetime(df1['published_at'])
df1['date'] = df1['datetime'].dt.date
df1['time'] = df1['datetime'].dt.time
sorted_df1 = df1.sort_values(by='likeCount', ascending=False).reset_index(drop=True)

cleaned_stats_df = df1.drop(columns=['published_at'])
cleaned_stats_df = cleaned_stats_df.drop(columns='datetime')
cleaned_stats_df.to_csv('../data/statistics.csv', index=False)
display(cleaned_stats_df)
```
