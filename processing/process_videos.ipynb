```python
import pandas as pd
import os
import json
```

```python
def read_first_json_file(directory_path):
    for filename in os.listdir(directory_path):
        if filename.endswith('.json'):
            filepath = os.path.join(directory_path, filename)
            with open(filepath, 'r', encoding='utf-8') as f:
                data = json.load(f)
            return data, filename
    return None, None

data, filename = read_first_json_file('../json_file')
if data:
    df = pd.DataFrame(data)
    df = df.drop(columns=['comments', 'statistics'])
    df.info()
```

```python
df['datetime'] = pd.to_datetime(df['published_at'])
df['date'] = df['datetime'].dt.date
df['time'] = df['datetime'].dt.time
cleaned_videos = df.drop(columns=['published_at', 'datetime'])
cleaned_videos.to_csv('../data/videos.csv', index=False)
cleaned_videos.info()
```
