```python
import sqlite3
import csv
import json 
import pandas as pd
```

```python
conn = sqlite3.connect('youtube.db')

cursor = conn.cursor()

videos_table = '''
        CREATE TABLE IF NOT EXISTS videos (
        video_id TEXT PRIMARY KEY,
        channel_title TEXT,
        title TEXT,
        description TEXT,
        date TEXT,
        time TEXT
        );
'''

comments_table = '''
        CREATE TABLE IF NOT EXISTS comments (
        comment_id INTEGER PRIMARY KEY AUTOINCREMENT,
        video_id TEXT NOT NULL,
        comment_text TEXT,
        author TEXT,
        date TEXT,
        time TEXT,
        FOREIGN KEY (video_id) REFERENCES videos(video_id)
        );
'''

statistics_table = '''
        CREATE TABLE IF NOT EXISTS statistics (
        statistic_id INTEGER PRIMARY KEY AUTOINCREMENT,
        video_id TEXT NOT NULL,
        channel_title TEXT,
        title TEXT,
        viewCount INTEGER,
        likeCount INTEGER,
        commentCount INTEGER,
        date TEXT,
        time TEXT,
        FOREIGN KEY (video_id) REFERENCES videos(video_id)
        );
'''

detail_comment = '''
        CREATE TABLE IF NOT EXISTS detail (
        detail_id INTEGER PRIMARY KEY AUTOINCREMENT,
        video_id TEXT NOT NULL,
        sent_comment TEXT,
        comment_text TEXT,
        author TEXT,
        date TEXT,
        time TEXT,
        FOREIGN KEY (video_id) REFERENCES videos(video_id)
        );
'''

sentiment_table = '''
        CREATE TABLE IF NOT EXISTS sentiments (
        sentiment_id INTEGER PRIMARY KEY AUTOINCREMENT,
        video_id TEXT NOT NULL,
        date TEXT,
        time TEXT,
        negative_pct REAL,
        neutral_pct REAL,
        positive_pct REAL,
        total_comments REAL,
        FOREIGN KEY (video_id) REFERENCES videos(video_id)
        );
'''
conn.commit()

cursor.execute(videos_table)
cursor.execute(comments_table)
cursor.execute(statistics_table)
cursor.execute(detail_comment)
cursor.execute(sentiment_table)
```

```python
df_videos = pd.read_csv('../data/videos.csv')
df_comments = pd.read_csv('../data/etc_comments.csv')
df_statistics = pd.read_csv('../data/statistics.csv')
df_detail_comment = pd.read_csv('../data/detail-comment.csv')
df_sentiment = pd.read_csv('../data/sentiment.csv')
```

```python
cursor.execute('DELETE FROM videos')
cursor.execute('DELETE FROM comments')
cursor.execute('DELETE FROM statistics')
cursor.execute('DELETE FROM detail')
cursor.execute('DELETE FROM sentiments')
conn.commit()
```

```python
df_videos.to_sql('videos', conn, if_exists='append', index=False)
df_comments.to_sql('comments', conn, if_exists='append', index=False)
df_statistics.to_sql('statistics', conn, if_exists='append', index=False)
df_detail_comment.to_sql('detail', conn, if_exists='append', index=False)
df_sentiment.to_sql('sentiments', conn, if_exists='append', index=False)
```
